### Making Own MBGD Regression Class From Scratch

```python
class MGDRegressor:
    def __init__(self,learning_rate,epochs,batch_size):
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        
    def fit(self,X_train,y_train):
        # initalizing intercept term to 0 and coeffiecients term to 1
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            for j in range(int(X_train.shape[0] / self.batch_size)):
                # selecting random row from X data 
                idx = np.random.choice(range(X_train.shape[0]),self.batch_size)
                # updating intercept terms
                # instead of whole data we taking subset
                y_hat = np.dot(X_train[idx],self.coef_) + self.intercept_
                intercept_der = -2 * np.mean(y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr * intercept_der)
                
                # updating / calculating Coefficient term
                coef_der = -2 * np.dot((y_train[idx] -y_hat),X_train[idx])
                self.coef_ = self.coef_ - (self.lr * coef_der)
        print(self.coef_,self.intercept_)
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
mgd = MGDRegressor(batch_size=int(X_train.shape[0] / 10),learning_rate=0.01,epochs=1000)
```

```python
mgd.fit(X_train,y_train)
```

```python
y_pred1 = mgd.predict(X_test)
```

```python
r2_score(y_test,y_pred1)
```
